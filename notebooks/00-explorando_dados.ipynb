# Verificação rápida dos dados

```python
from IPython.display import display
import duckdb

PARQUET_BRONZE_FILE_NAME = ["../data/02-bronze/classificacao_despesa.parquet",
                            "../data/02-bronze/cnpj.parquet",
                            "../data/02-bronze/despesa.parquet",
                            "../data/02-bronze/receita.parquet"
                            ]

PARQUET_SILVER_FILE_NAME = ["../data/03-silver/classificacao_despesa.parquet",
                            "../data/03-silver/cnpj.parquet",
                            "../data/03-silver/despesa.parquet",
                            "../data/03-silver/receita.parquet"
                            ]

PARQUET_GOLD_FILE_NAME = ["../data/04-gold/despesa_enriquecida.parquet",
                    "../data/04-gold/partido_ano_despesa.parquet",
                    "../data/04-gold/partido_ano_receita.parquet",
                    "../data/04-gold/receita_enriquecida.parquet"]

def verificar_dados(parquet_file_names):
    for file in parquet_file_names:
        print(f"\nArquivo: {file}")
    
        df = duckdb.sql(
            f"""
            SELECT *
            FROM read_parquet('{file}')
            USING SAMPLE 10 ROWS
            """
        ).df()
        print("\nShape:", df.shape)
        print("\nColunas:")
        print(df.dtypes)

        print("\nAmostra:")
        display(df)
```

## Visualiza todos os dados

```python
print("="*50, "\n\t Verificando arquivos BRONZE \n","="*50)
verificar_dados(PARQUET_BRONZE_FILE_NAME)

print("\n\n", "="*50, "\n\t Verificando arquivos SILVER \n", "="*50)
verificar_dados(PARQUET_SILVER_FILE_NAME)

print("\n\n", "="*50, "\n\t Verificando arquivos GOLD \n", "="*50)
verificar_dados(PARQUET_GOLD_FILE_NAME)
```
