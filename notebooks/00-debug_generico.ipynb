# Com DuckDB

```python
import duckdb
```

```python
PARQUET_FILE_NAME = "cnpj.parquet"
df = duckdb.sql(
    f"""
    SELECT
        *
    FROM read_parquet(
        '../data/02-bronze/{PARQUET_FILE_NAME}'
    )
    LIMIT 20
    """
).df()

df.head()
```
