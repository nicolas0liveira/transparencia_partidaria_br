# Com DuckDB

```python
import duckdb

df = duckdb.sql(
    """
    SELECT
        *
    FROM read_parquet(
        '../data/02-bronze/despesa.parquet'
    )
    LIMIT 20
    """
).df()

df.head()
```

```python
import duckdb

df = duckdb.sql(
    """
    SELECT
        sg_partido,
        SUM(vl_receita) AS vl_total
    FROM read_parquet(
        '../data/02-bronze/receita.parquet'
    )
    GROUP BY 1
    ORDER BY 2 DESC
    LIMIT 20
    """
).df()

df.head()
```
